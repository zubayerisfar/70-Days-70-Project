# **Preprocessing**

```python
original_news_df = train_news_df.copy()  # Backup 
train_news_df = train_news_df.sample(n=1000, random_state=42)
```

```python
train_news_df['Title'] = train_news_df['Title'].fillna('')
train_news_df['Abstract'] = train_news_df['Abstract'].fillna('')
```

```python
train_news_df['Combined_Features'] = train_news_df['Title'] + ' ' + train_news_df['Abstract'] + ' ' + train_news_df['Category'] + ' ' + train_news_df['SubCat']
```

```python
print(train_news_df['Combined_Features'].isna().sum())  
```

```python
train_news_df.drop(['NewsId','url','TitleEnt','AbstractEnt'], axis=1, inplace=True)
```

**Tokenization and Cleaning**

```python
stop_words = set(stopwords.words('english'))
def preprocess_text(text):
    tokens = word_tokenize(text.lower()) 
    tokens = [word for word in tokens if word not in string.punctuation]  
    tokens = [word for word in tokens if word not in stop_words]  
    return tokens

train_news_df['Processed_Combined_Features'] = train_news_df['Combined_Features'].apply(preprocess_text)
```

```python
def remove_entities(tokens):
    tagged_tokens = pos_tag(tokens)
    named_entities = ne_chunk(tagged_tokens)
    return [word for word, tag in tagged_tokens if not isinstance(tag, Tree)]  # Keep non-entity tokens

train_news_df['Processed_Combined_Features'] = train_news_df['Processed_Combined_Features'].apply(remove_entities)
```

**TF-IDF Vectorizer**

```python
tfidf = TfidfVectorizer(stop_words='english', min_df=0.001, max_df=0.999, ngram_range=(1, 3))
train_news_df['Processed_Combined_Features'] = train_news_df['Processed_Combined_Features'].apply(lambda x: ' '.join(x))

text_vectors = tfidf.fit_transform(train_news_df['Processed_Combined_Features'])
print(text_vectors.shape)
```

**Cosine Similarity**

```python
cosine_similarities = cosine_similarity(text_vectors, text_vectors)
```

# **Recommendation System and Accuracy**

```python
def calculate_cosine_similarity_accuracy(recommended_titles, relevant_titles):
    relevant_set = set(relevant_titles)
    
    # Number of recommended articles
    top_n = len(recommended_titles)
    
    # Calculate true positives 
    true_positives = len(set(recommended_titles) & relevant_set)
    
    # Precision
    precision = true_positives / top_n if top_n > 0 else 0
    
    # Recall
    recall = true_positives / len(relevant_set) if len(relevant_set) > 0 else 0
    
    # F1-Score
    if precision + recall > 0:
        f1_score = 2 * (precision * recall) / (precision + recall)
    else:
        f1_score = 0
    
    return precision, recall, f1_score, true_positives
```

**Based-Content-Filtering**

```python
category_input = input("Enter the category you are interested in: ")

filtered_by_category = train_news_df[train_news_df['Category'] == category_input]

if not filtered_by_category.empty:
    unique_subcategories = filtered_by_category['SubCat'].unique()
    print("\nAvailable subcategories:")
    for i, subcat in enumerate(unique_subcategories, start=1):
        print(f"{i}. {subcat}")
    
    subcat_choice = int(input("\nChoose the subcategory by entering the corresponding number: "))
    selected_subcategory = unique_subcategories[subcat_choice - 1]
    
    filtered_by_subcat = filtered_by_category[filtered_by_category['SubCat'] == selected_subcategory]
    
    if not filtered_by_subcat.empty:
        # Compute similarity scores using TF-IDF
        tfidf = TfidfVectorizer(stop_words='english')
        tfidf_matrix = tfidf.fit_transform(filtered_by_subcat['Combined_Features'])

        # Calculate pairwise cosine similarity
        similarity_matrix = cosine_similarity(tfidf_matrix)

        filtered_indices = list(filtered_by_subcat.index)  # using Map 
        recommended_titles = []
        relevant_titles = [] 
        for i, row_index in enumerate(filtered_indices):
            title = filtered_by_subcat.loc[row_index, 'Title']
            abstract = filtered_by_subcat.loc[row_index, 'Abstract']
            similarity_score = similarity_matrix[0, i]  
            print(f"Title: {title}")
            print(f"Abstract: {abstract}")
            print(f"Similarity Score: {similarity_score:.2f}")
            print("-" * 40)
            
            recommended_titles.append(title)
            relevant_titles.append(title)  #assuming all titles are relevant
        # acc
        precision, recall, f1_score, true_positives = calculate_cosine_similarity_accuracy(recommended_titles, relevant_titles)
        print(f"\nPrecision: {precision:.2f}")
        print(f"Recall: {recall:.2f}")
        print(f"F1-Score: {f1_score:.2f}")
        print(f"True Positives: {true_positives}")
    else:
        print("No articles found for the selected subcategory.")
else:
    print("No articles found for the selected category.")
```
